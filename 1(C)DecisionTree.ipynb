## 1. Import libraries and load MNIST

```python
import time, numpy as np, matplotlib.pyplot as plt, tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
np.random.seed(42); tf.random.set_seed(42)
(x_train, y_train), (x_test, y_test) = keras.datasets.mnist.load_data()
x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0
x_train = x_train.reshape(-1, 784); x_test = x_test.reshape(-1, 784)
print(x_train.shape, x_test.shape)
```

## 2. Helper functions

```python
def plot_history(history, title="Training curves"):
    fig, ax = plt.subplots(1, 2, figsize=(11, 4))
    ax[0].plot(history.history["loss"], label="train")
    if "val_loss" in history.history: ax[0].plot(history.history["val_loss"], label="validation")
    ax[0].set_title("Loss"); ax[0].legend()
    ax[1].plot(history.history["accuracy"], label="train")
    if "val_accuracy" in history.history: ax[1].plot(history.history["val_accuracy"], label="validation")
    ax[1].set_title("Accuracy"); ax[1].legend()
    fig.suptitle(title); plt.show()
```

## 3. Decision Tree Classifier

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, log_loss
model=DecisionTreeClassifier(max_depth=25, random_state=42)
start=time.time(); model.fit(x_train,y_train); train_time=time.time()-start
pred=model.predict(x_test); prob=model.predict_proba(x_test)
acc=accuracy_score(y_test,pred); loss=log_loss(y_test,prob,labels=np.arange(10))
print("Test accuracy:",acc,"Log loss:",loss,"Training seconds:",train_time)
```

## 4. Loss comparison\nA decision tree is not normally trained with an epoch-based loss history. This plot reports final train/test log loss.

```python
train_prob=model.predict_proba(x_train[:10000])
train_loss=log_loss(y_train[:10000],train_prob,labels=np.arange(10))
plt.plot(["Train sample","Test"],[train_loss,loss],marker="o")
plt.ylabel("Log loss"); plt.title("Final decision-tree log loss"); plt.show()
```
